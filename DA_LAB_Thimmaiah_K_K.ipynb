```python
import pandas as pd
from scipy.stats import ttest_1samp
data = pd.read_csv("/content/drive/MyDrive/python/InternetMobileTime.csv")
print("First five observations : ",data.head())
print("Column Names : ",data.columns)
sample=data["Minutes"]
n=len(sample)
sample_mean = sample.mean()
sample_std = sample.std()
print("\nSample Statistics")
print("-------------------------")
print("Sample size : ", n)
print("Sample mean : ", sample_mean)
print("Sample standard deviation :", sample_std)
hypothesized_mean = 144
alpha = 0.05
t_statistic, p_value = ttest_1samp(sample, hypothesized_mean)
print("\nHypothesis Test")
print("-------------------------")
print("Hypothesized mean : ", hypothesized_mean)
print("t-statistic : ", t_statistic)
print("p-value : ", p_value)
print("\nDecision")
print("-------------------------")
if p_value < alpha:
    print("Reject H0")
    print("There is sufficient evidence that the population mean is different from 144 minutes.")
else:
    print("Fail to reject")
    print("There is insufficient evidence that the population mean is different from 144 minutes.")
```

```python
import pandas as pd
import numpy as np
from scipy import stats


df = pd.read_csv("/content/drive/MyDrive/python/Luggage.csv")

print("Luggage Delivery Data")
print(df)

print("\nDescriptive Statistics")
print(df.describe())

mean_a = df["Wing A"].mean()
mean_b = df["Wing B"].mean()

print("\nMean Delivery Time")
print("Wing A:", mean_a)
print("Wing B:", mean_b)

sd_a = df["Wing A"].std()
sd_b = df["Wing B"].std()

print("\nStandard Deviation")
print("Wing A:", sd_a)
print("Wing B:", sd_b)


difference = mean_a - mean_b

print("\nDifference between Means:")
print(difference)

t_stat, p_value = stats.ttest_ind(
    df["Wing A"],
    df["Wing B"],
    equal_var=False
)

print("\nHypothesis Test Results")
print("t-statistic:", t_stat)
print("p-value:", p_value)


alpha = 0.05

if p_value < alpha:
    print("\nDecision: Reject H₀")
    print("There is a significant difference between")
    print("the mean delivery times of Wing A and Wing B.")
else:
    print("\nDecision: Fail to reject H₀")
    print("There is not enough evidence to conclude")
    print("that the mean delivery times are different.")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

data = [
    [15, 9],
    [16, 15]
]

table = pd.DataFrame(
    data,
    index=["Company A", "Company B"],
    columns=["Promoted", "Not Promoted"]
)

print("Observed Frequencies:")
print(table)

print("H0: Company and Promotion Status are independent.")
print("H1: Company and Promotion Status are not independent.")
print("Significance level (α) = 0.05")

chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    table,
    correction=False
)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", degrees_of_freedom)
print("p-value:", p_value)

expected_table = pd.DataFrame(
    expected,
    index=["Company A", "Company B"],
    columns=["Promoted", "Not Promoted"]
)

print("Expected Frequencies:")
print(expected_table)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is a significant association between")
    print("company and promotion status.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is insufficient evidence of an association")
    print("between company and promotion status.")

from scipy.stats import chi2 as chi2_dist

critical_value = chi2_dist.ppf(1 - alpha, degrees_of_freedom)

print("Critical value:", critical_value)
print("Calculated Chi-square:", chi2)

if chi2 > critical_value:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

```python
import pandas as pd
import scipy.stats as stats
data = pd.read_csv("/content/drive/MyDrive/python/AOVData_50_Vehicles.csv")
data.head()
data["fuel_type"].unique()
data.groupby("fuel_type")["co_emissions"].agg(["count", "mean", "std"])
petrol = data[data["fuel_type"] == "Petrol"]["co_emissions"]
e85 = data[data["fuel_type"] == "E85"]["co_emissions"]
diesel = data[data["fuel_type"] == "Diesel"]["co_emissions"]
cng = data[data["fuel_type"] == "CNG"]["co_emissions"]
F, p = stats.f_oneway(petrol, e85, diesel, cng)
print("F-statistic:", F)
print("p-value:", p)
if p < 0.05:
    print("Reject the null hypothesis")
    print("The mean carbon emissions are equal for all fuel types.")
else:
    print("Fail to reject the null hypothesis")
    print(" At least one fuel type has a different mean carbon emission. ")
```
